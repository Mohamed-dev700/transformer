```python
import torch
import torch.nn.functional as F

# Example sentence (7 tokens, embedding size = 4 for simplicity)
X = torch.tensor([
    [1.0, 0.0, 1.0, 0.0],   # What
    [0.0, 2.0, 0.0, 2.0],   # are
    [1.0, 1.0, 1.0, 1.0],   # the
    [2.0, 1.0, 2.0, 1.0],   # symptoms
    [0.0, 1.0, 0.0, 1.0],   # of
    [1.0, 2.0, 1.0, 2.0],   # diabetes
    [0.5, 0.5, 0.5, 0.5]    # ?
])

# Define weight matrices
d_k = X.shape[1]

W_Q = torch.rand(d_k, d_k)
W_K = torch.rand(d_k, d_k)
W_V = torch.rand(d_k, d_k)

# Compute Q, K, V
Q = X @ W_Q
K = X @ W_K
V = X @ W_V

# Compute attention scores
scores = Q @ K.T / torch.sqrt(torch.tensor(d_k, dtype=torch.float32))

# Apply softmax
attention_weights = F.softmax(scores, dim=-1)

# Compute output
output = attention_weights @ V

print("Attention Weights:\n", attention_weights)
print("\nOutput:\n", output)
```
